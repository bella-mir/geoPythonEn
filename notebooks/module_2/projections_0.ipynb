# II. Projections and Coordinate Reference Systems


## Topics


1. [**Introduction to Map Projections and Coordinate Reference Systems**](projections_1.ipynb)

   An introduction to the two main types of coordinate reference systems — geographic and projected. You will learn what a CRS is, why it matters, and how to inspect CRS information in a spatial dataset.

2. [**(Re)projecting Spatial Data**](projections_2.ipynb)

   You will learn how to change the coordinate reference system of a spatial dataset using GeoPandas, and how to identify the appropriate CRS for your analysis.

3. [**Measuring Areas and Distances**](projections_3.ipynb)

   You will learn how the choice of CRS affects spatial measurements, and how to calculate distances and areas in metres using projected data.

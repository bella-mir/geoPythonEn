# Introduction to Map Projections and Coordinate Reference Systems


In this section, we will look at what map projections and coordinate reference systems are, and how the choice of projection affects the results of spatial analysis and data visualisation.


## 1. Background Theory


Although the Earth is commonly **depicted as a sphere**, this is only a **simplified model**. In geodesy, a more accurate representation is the **geoid** — a surface of equal gravitational potential that approximately coincides with mean sea level and extends beneath the continents.

![Geoid](images/geoid.jpg)

_Source: [Wikimedia Commons](https://commons.wikimedia.org/wiki/File:Geoid_undulation_to_scale.jpg)_

Because the geoid has an irregular shape, cartography typically approximates it with a simpler mathematical model — an **ellipsoid**.

An ellipsoid can be thought of as a **slightly flattened sphere**. This shape is far easier to describe mathematically while still providing a close approximation of the Earth's true form.

The Earth is not a perfect sphere. Its equatorial radius is approximately **6,378 km**, while its polar radius is around **6,357 km** — a difference of about **21 km**, meaning the Earth is slightly **flattened at the poles and bulges at the equator**.

![Ellipsoid](images/oblateearth.png)

_Own figure, generated by `images/oblateearth.py`._

In geodesy and cartography, a **reference ellipsoid** is used — a specifically chosen ellipsoid whose parameters are fitted to best represent the average shape of the Earth. This ellipsoid serves as the **mathematical foundation for coordinate reference systems**.

The positions of points on the Earth's surface — latitude, longitude, and elevation — are defined **relative to the reference ellipsoid**, not relative to the Earth's actual physical surface.

### 1.1. Map Projections

<span style="color:#03A3A6">A map projection is a mathematically defined method of representing the Earth's surface on a flat plane.</span>

Transferring coordinates from the ellipsoid onto a map inevitably introduces **distortions**.

No projection can simultaneously preserve **all geometric properties** of features. Depending on the projection, any of the following may be distorted:

- **angles and shapes**
- **areas**
- **distances**
- **directions**

Different projections **distribute these distortions differently**, which is why each one is better suited to certain tasks.

#### 1.1.1. Types of Projections

##### **By preserved property**

One way to classify projections is by **which properties they preserve**.

Cartography generally distinguishes several types:

- **conformal** — preserve angles and local shapes;
- **equal-area** — preserve areas;
- **equidistant** — preserve distances along certain lines or directions;
- **compromise** — do not fully preserve any single property, but aim to minimise overall distortion.

##### **By developable surface**

Projections are also classified by the type of **developable surface** used to transfer coordinates from the ellipsoid onto the flat map plane.

The most common types are:

- **cylindrical projections**
- **conic projections**
- **azimuthal (planar) projections**

![Projections_Developable_Surface](images/map-projections.jpg)

_Source: [UCGIS.ORG](https://gistbok-topics.ucgis.org/CV-03-006)_

Each group **distributes distortions differently**, making each one better suited to different **latitudes, territory shapes, and map scales**.

##### **By orientation of the developable surface**

In addition to the type of developable surface, projections also differ by its **orientation relative to the Earth**.

Three main cases are distinguished:

- **normal** — the axis of the surface aligns with the Earth's axis of rotation;
- **transverse** — the surface is rotated 90° relative to the Earth's axis;
- **oblique** — the surface is set at an arbitrary angle.

![Projections_Orientation](images/proj_aspekt.jpg)

_Source: after Furuti, 2002._

The orientation of the projection determines **where distortions on the map will be smallest**.


#### 1.1.2. Choosing a Map Projection

Selecting an appropriate map projection is an important step when preparing spatial data for analysis or visualisation. Since projecting the Earth's surface onto a flat plane always introduces distortions, different projections are **optimised for different tasks and territory sizes**.

Several factors typically guide the choice of projection:

- **the type of analysis or map** being produced;
- **the geographic location of the area** (latitude, spatial extent);
- **the scale** — local, regional, or global;
- **the properties that need to be preserved** (areas, shapes, distances, or directions).

For example:

- **equal-area projections** are used when accurate comparison of **areas** is important;
- **conformal projections** are used when preserving **shapes and angles** matters — for example in navigation or directional analysis;
- **equidistant projections** are used when accurate representation of **distances along specific directions** is required;
- **compromise projections** are often used for **world maps**, where the goal is to reduce overall distortion without strictly preserving any single property.

The choice of projection is especially important when creating **thematic maps**, where spatial distortions can affect how data is interpreted. For example, when displaying statistical data on **choropleth maps**, it is recommended to use an **equal-area projection** so that region sizes are represented correctly.

#### 1.1.3. The Mercator Projection

The **Mercator projection** is a cylindrical conformal map projection introduced by the Flemish cartographer Gerardus Mercator in 1569.

![Mercator Projection](images/mercator.png)

_Source: [Université Lyon 1 — Cylindrical Projections](https://math.univ-lyon1.fr/~alachal/diaporamas/diaporama_cartographie3/Cylindrical_Projections.html)_

Its defining property is that it **preserves angles and local shapes**. As a result, directions on the map correspond to directions on the ground, which made the Mercator projection especially well suited to **maritime navigation**. Sailors could plot a course as a straight line on the map and follow that constant bearing using a compass.

However, the Mercator projection significantly **distorts areas**, especially at high latitudes. As you move towards the poles, the apparent size of landmasses increases dramatically. For example, Greenland can appear comparable in size to Africa on a Mercator map, even though Africa is roughly **14 times larger**.

![The true size](images/Worlds_animate.gif)

_Source: [Wikimedia Commons](https://commons.wikimedia.org/wiki/File:Worlds_animate.gif)_

> The interactive tool at https://thetruesize.com lets you drag country outlines across the map and compare their true sizes — vividly illustrating how severely **areas can be distorted in the Mercator projection**.

To reduce distortions when working with smaller areas, modified versions of the Mercator projection are used. One of the most widely adopted is the **Universal Transverse Mercator (UTM)**.

#### 1.1.4. Universal Transverse Mercator (UTM)

**Universal Transverse Mercator (UTM)** is one of the most widely used projection systems in GIS for working with data at local and regional scales.

It is based on the **transverse Mercator projection**. The UTM system divides the Earth's surface into **60 longitudinal zones**, each **6° of longitude** wide. Each zone has its own projection centred on a central meridian, relative to which the coordinate transformation is applied.

Separate coordinate systems are defined for the northern and southern hemispheres, giving a total of **120 UTM variants**: 60 zones per hemisphere.

Dividing the Earth into relatively narrow zones greatly **reduces projection distortions within each zone**, making UTM particularly well suited for analysing data **within a single zone** — for example, at the scale of a **city, region, or small country**.

### 1.2. Coordinate Reference Systems

<span style="color:#03A3A6">**A coordinate reference system** (_CRS — Coordinate Reference System_) is a mathematical framework that allows the position of points on the Earth's surface to be described using coordinates.</span>

#### 1.2.1. Geographic Coordinate Systems

Geographic coordinate systems are familiar from school geography. In these systems, the position of a point on the Earth's surface is defined by **latitude** and **longitude**.

Coordinates are measured in **degrees** and are defined relative to:

- the **equator** — for latitude
- the **prime meridian** — for longitude

One of the most widely used geographic coordinate systems is **WGS 84**.

It is the system used in **GPS navigation** and in many open **spatial datasets**.

> Because coordinates in geographic systems are expressed in degrees, these systems are not always suitable for measuring distances or areas. For such tasks, **projected coordinate systems** are used.


> **Interesting fact**
>
> The Greenwich meridian was officially adopted as the prime meridian in 1884 at the International Meridian Conference in Washington, D.C.

#### 1.2.2. Projected Coordinate Systems

In **projected coordinate systems**, the position of a point is defined on a **flat map plane** using a Cartesian coordinate system.

Each point is described by two coordinates — **X** and **Y** — measured from an origin point.

These coordinates are obtained through a **map projection** — the mathematical transformation that converts positions on the Earth's surface (latitude and longitude) onto a flat map plane.

Unlike geographic coordinates, the values of **X** and **Y** cannot be interpreted meaningfully without knowing the coordinate system they belong to.

In geographic coordinate systems, positions are expressed in **degrees** (latitude and longitude), while projected systems use **linear units** — most commonly **metres**.

![Coordinate Systems](images/coordinateSystem.png)

_Source: [Esri](https://www.esri.com/)_

### 1.3. EPSG Codes

**EPSG codes** are numeric identifiers used to refer to **coordinate reference systems, map projections, and related coordinate system components**. The acronym stands for the _European Petroleum Survey Group_, the organisation that originally developed the registry.

Initially created for use in the oil and gas industry, the EPSG registry has since become a widely adopted standard and is now used throughout **geographic information systems (GIS)**.

Each coordinate system or projection is assigned a **unique numeric code**.

For example:

- the widely used geographic coordinate system **WGS 84** has the code **EPSG:4326**

For the **UTM** system, EPSG codes follow this pattern:

- **EPSG:326xx** — UTM zones in the **northern hemisphere**
- **EPSG:327xx** — UTM zones in the **southern hemisphere**

where **xx** is the UTM zone number.


## 2. The CRS of a GeoDataFrame


The geometries in a `GeoDataFrame` are stored in a specific **coordinate reference system**.

This information is held in the **`.crs`** attribute.

Before starting any spatial analysis, it is important to check:

- whether a CRS is defined for the data
- what type of CRS it is
- whether the units of the CRS are appropriate for the analysis being performed
- whether all datasets being used share the same CRS


### 2.0. Preparing the Data


Let's load the district boundary from OpenStreetMap.


```python
import osmnx as ox

# keep all downloaded OSM responses in one cache folder at the root of the repository
ox.settings.cache_folder = "../../cache"

area_name = "Tsentralny District, Saint Petersburg"

admin_border = ox.geocode_to_gdf(area_name)
admin_border.explore(tiles="cartodbpositron")
```

### 2.1. General CRS Information


To inspect the coordinate reference system of the data, we use the `.crs` attribute.


```python
admin_border.crs
```

We can see that the data is in **EPSG:4326 (WGS 84)**.

As mentioned earlier, this is **one of the most widely used geographic coordinate systems**, adopted by the majority of global spatial datasets and GPS navigation systems.

Key properties of this CRS:

- **type**: geographic coordinate system
- **axes**: latitude and longitude
- **units**: degrees
- **area of use**: the entire world
- **ellipsoid**: WGS 84
- **prime meridian**: Greenwich


Each of these properties can also be retrieved individually. Let's look at how to do that.


### 2.2. CRS Type

One of the key properties of a coordinate system is its type — whether coordinates are expressed in degrees (a geographic CRS) or in linear units such as metres (a projected CRS).

The CRS type can be checked using the `.is_geographic` and `.is_projected` attributes.


```python
print(f"Geographic: {admin_border.crs.is_geographic}")
print(f"Projected: {admin_border.crs.is_projected}")
```

### 2.3. EPSG Code

The EPSG code of the coordinate reference system can be retrieved using the `.crs.to_epsg()` method.

```python
admin_border.crs.to_epsg()
```

### 2.4. Coordinate Axes

The CRS also holds information about the coordinate axes and the units in which coordinates are expressed.


```python
admin_border.crs.axis_info
```

The fields shown are:

- **name** — the name of the coordinate axis
- **abbrev** — the abbreviated axis label
- **direction** — the direction in which the coordinate value increases
- **unit_name** — the unit of measurement

In this case, two axes are listed:

- **Geodetic latitude (Lat)** — direction: **north**, units: **degrees**
- **Geodetic longitude (Lon)** — direction: **east**, units: **degrees**


### 2.5. Area of Use

Every coordinate reference system has an **area of use** — the region for which it is intended.

Coordinate systems are typically designed **for specific regions** in order to minimise distortions of coordinates, distances, and areas within that territory.

When working with spatial data, it is therefore important to ensure that the CRS being used **is appropriate for the region being analysed**.


```python
admin_border.crs.area_of_use
```

This shows the bounding coordinates of the territory for which the CRS is intended:

- **west** — western boundary (minimum longitude)
- **south** — southern boundary (minimum latitude)
- **east** — eastern boundary (maximum longitude)
- **north** — northern boundary (maximum latitude)
- **name** — the name of the area of use

**WGS 84 (EPSG:4326)** is intended for use **worldwide** — as expected, since it is a global geographic coordinate system used in GPS and the vast majority of global spatial datasets.

_Many projected coordinate systems are designed only for specific regions. We will return to this in the following sections._


> **Checking the coordinate reference system** is a critical step when working with spatial data.
> The CRS affects not only how data is displayed on a map, but also the accuracy of spatial calculations — such as measuring **distances**, **areas**, and constructing **buffer zones**.
>
> If the CRS is incorrect, features may be **mislocated in space**. Furthermore, if the datasets being used are in **different coordinate systems**, spatial operations may produce incorrect results.
>
> Before starting any analysis, it is therefore important to **verify the CRS of your data** and, where necessary, reproject all datasets **to a common CRS**.

# I. Spatial Data

## Topics


1. [**Spatial Data Models**](spData_1.ipynb)

   An introduction to the two main ways of representing spatial data – vector and raster. You will learn how information is stored in each model and which Python classes are used to work with them.

2. [**Reading Vector Data**](spData_2.ipynb)

   You will learn how to read vector data from a variety of formats, create a spatial dataset from coordinates, and save the result.

3. [**Exploring a Dataset**](spData_3.ipynb)

   You will learn how to retrieve key information about a dataset: its coordinate reference system, feature count, geometry type, attribute columns, data types, and more.

4. [**Exporting Data from OSM**](spData_4.ipynb)

   You will learn how to download data from OpenStreetMap by tag and value using the `osmnx` library.


## Assignment: Collecting the Data

Each module ends with a hands-on assignment – a stage of the **"15-Minute City"** project, which looks at how accessible everyday amenities are. You can carry out the project for any city of your choice.

In [this module's assignment](spData_task.ipynb) you prepare the data the rest of the project runs on: you choose a study area, download its boundary and several categories of everyday amenities from OpenStreetMap, clean the results, and save them as a GeoPackage with one layer per category.

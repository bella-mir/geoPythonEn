# VI. Interactive Visualisation


## Topics

1. [**Web Map with Folium**](map_1.ipynb)

   Learn how to build web maps with Folium: add multiple layers (land use, stops, service areas), configure styles, tooltips, and map controls.

2. [**Publishing on GitHub Pages**](map_2.ipynb)

   Learn how to save a map as HTML and host it online with GitHub Pages so you can share your analysis results via a link.


## Assignment: An Interactive Map of the Results

Each module ends with a hands-on assignment — a stage of the **"15-Minute City"** project, which looks at how accessible everyday amenities are. You can carry out the project for any city of your choice.

In [this module's assignment](map_task.ipynb) you bring every layer of the project onto a single Folium map — the study area, land use, the amenities and stops, and their catchment areas — with layer groups, a legend and map controls, and save the result as an HTML file.

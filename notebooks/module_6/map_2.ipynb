# Publishing on GitHub Pages

In this notebook we'll walk through how to publish an interactive map saved as an `.html` file using **GitHub Pages**.

GitHub Pages lets you host static web pages directly from a GitHub repository, for free. Since a Folium map is saved as a plain HTML file, it can be published as a web page and opened via a URL in any browser.


## What You'll Need

To publish the map you will need:

- a GitHub account;
- an HTML file containing your Folium map;
- a GitHub repository;
- GitHub Pages enabled.

The result will be a URL of the form:

```text
https://username.github.io/repository-name/
```


## 1. Save the Folium Map to HTML

After building the Folium map, save it to a file.

If the map is stored in variable `m`, use the `.save()` method:

```python
m.save("index.html")
```

It's important to name the file `index.html`, because GitHub Pages serves this file as the site's home page by default.

If you save it under a different name, e.g.:

```python
m.save("map.html")
```

the URL will look like this:

```text
https://username.github.io/repository-name/map.html
```

For simplicity, `index.html` is the recommended name.


## 2. Test the HTML File Locally

Before publishing, verify that the map opens correctly on your machine.

Find the `index.html` file in your project folder and open it in a browser.

If the map renders correctly, the file is ready to publish.

One thing to be aware of: your data is embedded in the file, but the **basemap tiles are not** — they are fetched from a tile server as the map is viewed. Open the file without an internet connection and the data will be there while the background is blank. That is expected, and it applies to the published version too: the page needs a connection to draw its basemap.


## 3. Create a GitHub Repository

Open GitHub and create a new repository.

For example, you might name it:

```text
folium-map
```

When creating the repository, you can choose:

- `Public` — if the map is meant to be shared with everyone;
- `Private` — if the repository should be restricted.

For a free GitHub account, Pages works only from **public** repositories: publishing a site from a private repository requires a paid plan. If the map is meant to be shared, a public repository is the simplest route.


## 4. Upload the Map File via the GitHub Interface

The simplest approach is to upload the file through the GitHub web interface.

Open the repository you just created and click:

```text
Add file → Upload files
```

Upload the file:

```text
index.html
```

Then click:

```text
Commit changes
```

The HTML file is now in the repository.


## 5. Enable GitHub Pages

Open the repository settings:

```text
Settings → Pages
```

Under **Build and deployment**, select:

```text
Source: Deploy from a branch
```

Then set:

```text
Branch: main
Folder: / (root)
```

Click:

```text
Save
```

GitHub will begin deploying the site.

After a short wait, a link to the published map will appear on this same page.

It will look something like:

```text
https://username.github.io/folium-map/
```


## 6. Open the Published Map

Navigate to the URL shown by GitHub Pages in the settings.

If the file is named `index.html`, the map will open directly at the repository URL:

```text
https://username.github.io/folium-map/
```

If the file has a different name, e.g. `map.html`, append the filename to the URL:

```text
https://username.github.io/folium-map/map.html
```


## 7. Updating the Map

If you modify the map in Python, save it again:

```python
m.save("index.html")
```

Then upload the new `index.html` to replace the old one.

Through the GitHub interface:

1. Open `index.html` in the repository;
2. Click the pencil icon to edit, or upload a new file via `Upload files`;
3. Save the changes with `Commit changes`.

After updating, GitHub Pages may take a moment to apply the changes. If the old version is still showing, refresh the page or open the link in a private/incognito window.


## 8. Publishing via Git

If you work with Git locally, you can push the map to the repository from the terminal.

Navigate to your project folder:

```bash
cd path/to/your/project
```

Initialize Git if it hasn't been initialized yet:

```bash
git init
```

Add the map file:

```bash
git add index.html
```

Create a commit:

```bash
git commit -m "Add Folium map"
```

Connect the remote repository:

```bash
git remote add origin https://github.com/username/folium-map.git
```

Push the files to GitHub:

```bash
git branch -M main
git push -u origin main
```

Then enable GitHub Pages in the repository settings as described above.


## 9. If the Map Uses External Files

Folium typically saves a self-contained HTML file, but some projects may depend on additional files.

In that case, upload not only `index.html` but all associated files and folders to the repository.

For example:

```text
folium-map/
├── index.html
├── data/
│   └── stops.geojson
└── photos/
    ├── stop_1.jpeg
    └── stop_2.jpeg
```

File paths in the HTML must match the folder structure in the repository.


## Summary

In this notebook we walked through how to publish an interactive Folium map on GitHub Pages.

We:

- saved the map to `index.html`;
- tested the HTML file locally;
- created a GitHub repository;
- uploaded the map to the repository;
- enabled GitHub Pages;
- obtained a public URL for the map.

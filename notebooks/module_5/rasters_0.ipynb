# V. Raster Analysis

## Topics

1. [**Raster Data Format**](rasters_1.ipynb)

   Learn how raster data is structured and how to work with it: reading, visualising, processing, and preparing it for analysis. Demonstrated using a WorldPop population distribution raster for Austria.

2. [**Zonal Statistics**](rasters_2.ipynb)

   Learn how to aggregate raster data by vector zones and compute metrics (such as population density) for territorial analysis.

## Assignment: How Much of the Population Each Amenity Serves

Each module ends with a hands-on assignment — a stage of the **"15-Minute City"** project, which looks at how accessible everyday amenities are. You can carry out the project for any city of your choice.

In [this module's assignment](geopy_task5.ipynb) you use zonal statistics over a WorldPop raster to count how many people live inside the catchment areas built in the previous assignment — and how many are left outside — then compare the coverage across all the categories of amenities.

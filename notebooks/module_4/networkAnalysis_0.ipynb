# IV. Fundamentals of Network Analysis


## Topics


1. [**The Street Network Graph**](networkAnalysis_1.ipynb)

   An introduction to representing a street network as a graph: retrieving one from OpenStreetMap, reading its structure of nodes and edges, and checking its degree distribution and connectivity.

2. [**Street Network Analysis**](networkAnalysis_2.ipynb)

   You will learn the core network analysis tasks: finding shortest paths along the street network, measuring the centrality of its nodes, building a distance matrix between features, and constructing service areas reachable from a point.

3. [**Network Analysis with External APIs**](networkAnalysis_3.ipynb)

   The same tasks handed to external routing services — OSRM, OpenRouteService and GraphHopper — instead of a local graph, and how to work with a web API along the way: building the request, checking the response, and turning the result into geometry.


## Assignment: Walking Isochrones

Each module ends with a hands-on assignment — a stage of the **"15-Minute City"** project, which looks at how accessible everyday amenities are. You can carry out the project for any city of your choice.

In [this module's assignment](networkAnalysis_task.ipynb) you build 5, 10 and 15-minute walking catchment areas on a local street network graph for one category of amenities, wrap the calculation into a function, and apply it to the other categories prepared in the first assignment.

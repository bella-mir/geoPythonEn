# III. Core Geoprocessing Tools


## Topics


1. [**Basic Geometry Operations**](geoprocessing_1.ipynb)

   An introduction to the core geoprocessing tools for working with geometry – buffer, dissolve, and clip.

2. [**Spatial Relationships**](geoprocessing_2.ipynb)

   You will learn how to determine spatial relationships between features using predicates such as `intersects`, `contains`, and `within`, and how to filter data by location.

3. [**Spatial Join**](geoprocessing_3.ipynb)

   You will learn how to join spatial datasets by containment using `sjoin` and by proximity using `sjoin_nearest`, based on the spatial relationships between features.

4. [**Aggregating Point Data with a Regular Grid**](geoprocessing_4.ipynb)

   You will build a regular grid and count the number of features within each cell.

5. [**Aggregating Attribute Data**](geoprocessing_5.ipynb)

   You will learn how to compute summary statistics from the attributes of spatially joined data.


## Assignment: Access to Public Transport

Each module ends with a hands-on assignment – a stage of the **"15-Minute City"** project, which looks at how accessible everyday amenities are. You can carry out the project for any city of your choice.

In [this module's assignment](geoprocessing_task.ipynb) you find out which buildings in your study area lie within walking distance of a public transport stop: you build catchment areas around the stops, join the buildings to them, count how many are left outside, and use the land use data from the previous assignment to estimate which of those are residential.
